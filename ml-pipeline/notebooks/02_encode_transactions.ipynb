# 02 — Encode Transactions

Loads each persona's raw transaction JSON and converts every transaction to a
discrete token using `encoder.py`. Output written to `output/tokens_{persona}.json`.

```python
import json
import sys
from pathlib import Path

sys.path.insert(0, str(Path('..') / 'src'))
from encoder import encode_persona, encode_transaction

OUTPUT = Path('..') / 'output'
PERSONAS = ['aisyah', 'daniel', 'mei_ling', 'hafiz']
```

```python
for persona in PERSONAS:
    with open(OUTPUT / f'{persona}.json', encoding='utf-8') as f:
        transactions = json.load(f)

    tokens = encode_persona(transactions)

    out_path = OUTPUT / f'tokens_{persona}.json'
    with open(out_path, 'w', encoding='utf-8') as f:
        json.dump(tokens, f, indent=2)

    unique = len(set(tokens))
    print(f'{persona:<12} {len(tokens):>4} tokens  {unique:>3} unique types')
```

```python
# Spot-check: show first 10 tokens for each persona
for persona in PERSONAS:
    with open(OUTPUT / f'tokens_{persona}.json', encoding='utf-8') as f:
        tokens = json.load(f)
    print(f'\n{persona}')
    for t in tokens[:10]:
        print(f'  {t}')
```
